# Fase 1 – Definición del problema

**Contexto.** Una empresa de telecomunicaciones pierde ingresos por clientes que cancelan su servicio (*churn*). Retener a un cliente cuesta bastante menos que captar uno nuevo.

**Objetivo general.** Identificar a los clientes con alto riesgo de abandono para dirigir campañas de retención.

**Objetivos específicos.**
1. Segmentar a los clientes según su comportamiento de consumo (modelo descriptivo).
2. Construir y comparar modelos que predigan el abandono (modelos predictivos).
3. Determinar qué factores influyen más en la decisión de abandonar.

**Alcance.** Datos históricos del dataset *Telco Customer Churn* (IBM). No incluye despliegue en producción.

**Criterio de éxito.** Priorizar el *recall* de la clase "Churn": un abandono no detectado cuesta más que contactar a un cliente que no se iba.
