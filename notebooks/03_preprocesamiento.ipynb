## Generación de variables
- `NumServicios`: cantidad de servicios contratados.
- `CargoPromedioMensual`: TotalCharges / tenure.
- `GrupoAntiguedad`: tramos de antigüedad.
- `ContratoLargo`, `PagoAutomatico`: indicadores binarios.

```python
df = generar_variables(df)
df[['tenure','NumServicios','CargoPromedioMensual','GrupoAntiguedad','ContratoLargo','PagoAutomatico']].head()
```

```python
X, y = separar_xy(df)
X_train, X_test, y_train, y_test = dividir(X, y)
prep = construir_preprocesador(X_train)
Xt = prep.fit_transform(X_train)
print('Entrenamiento:', X_train.shape, '-> transformado:', Xt.shape)
print('Proporción churn train/test:', round(y_train.mean(),3), round(y_test.mean(),3))
```
